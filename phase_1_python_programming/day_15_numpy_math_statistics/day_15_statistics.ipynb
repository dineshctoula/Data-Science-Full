### 📖 Study Notes

#### 1. Basic Statistical Functions
- NumPy provides built-in statistical functions that work efficiently on arrays:
  - `np.sum(arr)`: Sums all elements.
  - `np.mean(arr)`: Computes the arithmetic mean.
  - `np.median(arr)`: Finds the median value.
  - `np.min(arr)` / `np.max(arr)`: Finds minimum / maximum values.
  - `np.std(arr)` / `np.var(arr)`: Computes standard deviation and variance.

#### 2. Operations Along Axes
- On 2D or multi-dimensional arrays, you can perform computations along specific directions using the `axis` parameter:
  - `axis=0`: Operations along columns (reduces rows).
  - `axis=1`: Operations along rows (reduces columns).

#### 3. Searching & Logical Functions
- **`np.argmin()` / `np.argmax()`**: Return the index of the minimum or maximum value in the array.
- **`np.where(condition, [x, y])`**:
  - If only `condition` is passed, returns the indices of elements where the condition is True.
  - If `x` and `y` are also passed, returns elements from `x` where the condition is True and elements from `y` where it is False.

#### Exercise 1: Basic Statistics
Generate a random array of 100 values from a normal distribution with mean 10 and standard deviation 2. Compute the sample mean and sample standard deviation to verify they are close to the population parameters.

```python
np.random.seed(42)
samples = np.random.normal(loc=10, scale=2, size=100)

print("Sample Mean:", np.mean(samples))
print("Sample Standard Deviation:", np.std(samples))
```

#### Exercise 2: Searching and Filtering with `np.where`
Given an array of daily temperatures, find the index of the hottest day, and mark days with temperature > 25°C as 'Hot' and others as 'Normal'.

```python
temps = np.array([21.5, 24.3, 27.8, 19.2, 28.5, 23.0])

hottest_day_idx = np.argmax(temps)
print(f"Hottest day index: {hottest_day_idx} (temp: {temps[hottest_day_idx]}°C)")

categories = np.where(temps > 25, "Hot", "Normal")
print("Classifications:", categories)
```

#### Exercise 3: Practical Challenge (Student Exam Grades Analysis)
Create a 4x4 matrix representing exam grades (students as rows, subjects as columns). Find the average score for each subject, and the top-performing student index.

```python
# Rows: 4 students, Columns: Math, Science, English, History
grades = np.array([
    [85, 90, 78, 92],
    [72, 85, 88, 80],
    [95, 92, 96, 98],
    [68, 70, 75, 72]
])

# Calculate average per subject (column-wise)
subject_avg = np.mean(grades, axis=0)
print("Average grades per subject (Math, Science, English, History):")
print(np.round(subject_avg, 2))

# Calculate average per student (row-wise)
student_avg = np.mean(grades, axis=1)
print("\nAverage grades per student:")
print(np.round(student_avg, 2))

# Locate the topper
topper_idx = np.argmax(student_avg)
print(f"\nTop performing student index: {topper_idx} (Average: {student_avg[topper_idx]:.2f}%)")
```
