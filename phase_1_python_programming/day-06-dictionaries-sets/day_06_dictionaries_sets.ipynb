### 📖 Topics Overview
Today, we are shifting our focus to **Dictionaries** and **Sets**, two of the most widely used and powerful built-in data structures in Python for Data Science.

1. **Dictionaries (`dict`)**:
   - Unordered (until Python 3.7+ where insertion order is preserved), mutable collections of key-value pairs.
   - Extremely fast lookup times (O(1) average case) because they are implemented as hash tables.
   - Keys must be of an **immutable** type (strings, numbers, or tuples) and must be unique.
2. **Sets (`set`)**:
   - Unordered, mutable collections of **unique** elements.
   - Commonly used for membership testing, removing duplicates, and mathematical set operations.

### 1. Dictionaries: Basics, Syntax, and Manipulation
Let's learn how to define dictionaries, access values directly and safely, and modify elements.

```python
# 1. Creation
student = {
    "name": "Dinesh",
    "age": 22,
    "courses": ["Python", "Data Science"],
    "is_active": True
}
print("Initial student:", student)

# 2. Accessing values safely vs directly
print("Name:", student["name"])

# Using .get() prevents KeyError if the key doesn't exist
print("Email (using get):", student.get("email"))
print("Email (with default value):", student.get("email", "no-email@example.com"))

# 3. Modifying & Adding items
student["age"] = 23  # Modify
student["email"] = "dinesh@example.com"  # Add new
print("Updated student:", student)
```

### 2. Dictionary Methods & Iteration
Python dictionaries provide various built-in methods for retrieving views of keys, values, items, or deleting entries.

```python
# Get keys, values, and items
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))

# Iteration
print("\nIterating through keys and values:")
for key, value in student.items():
    print(f"Key: {key:<10} | Value: {value}")

# Deletion using pop()
removed_val = student.pop("is_active")
print(f"\nRemoved 'is_active' (value: {removed_val})")

# Updating dictionary inline using .update()
student.update({"age": 24, "location": "Kathmandu"})
print("After .update() method:", student)
```

### 3. Sets: Uniqueness & Basics
A set contains unique, unordered elements. Let's see how sets automatically filter duplicates, and how we add or remove elements.

```python
# Set creation automatically filters duplicates
numbers_list = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers = set(numbers_list)
print("Original list:", numbers_list)
print("Set (unique elements):", unique_numbers)

# Adding elements
unique_numbers.add(6)
print("Set after add(6):", unique_numbers)

# Removing elements (remove vs discard)
unique_numbers.remove(2)  # Will raise KeyError if 2 is not present
print("Set after remove(2):", unique_numbers)

unique_numbers.discard(10)  # Safe: does not raise error if 10 is missing
print("Set after discard(10):", unique_numbers)
```

### 4. Set Operations
Sets support mathematical operations like **Union**, **Intersection**, **Difference**, and **Symmetric Difference**.

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)

# Union (All unique elements from both sets)
print("Union (A | B):", set_a | set_b)

# Intersection (Elements common to both sets)
print("Intersection (A & B):", set_a & set_b)

# Difference (Elements in A but not in B)
print("Difference (A - B):", set_a - set_b)

# Symmetric Difference (Elements in A or B, but not both)
print("Symmetric Difference (A ^ B):", set_a ^ set_b)
```

### 5. Practical Exercise: Word Frequency Counter
**Objective**: Write a program that counts the occurrences of each word in a given paragraph and stores the counts in a dictionary.

We need to:
1. Clean basic punctuation from the paragraph.
2. Standardize to lowercase so capitalization doesn't create duplicate counts.
3. Count words using a dictionary.

```python
import string

def count_words(paragraph):
    # Remove punctuation
    cleaned_text = ""
    for char in paragraph:
        if char not in string.punctuation:
            cleaned_text += char
        else:
            cleaned_text += " "  # Replace punctuation with spaces
            
    # Lowercase and split
    words = cleaned_text.lower().split()
    
    # Count frequencies
    frequencies = {}
    for word in words:
        frequencies[word] = frequencies.get(word, 0) + 1
        
    return frequencies

# Sample Paragraph
sample_paragraph = (
    "Data science is an interdisciplinary academic field that uses statistics, "
    "scientific computing, scientific methods, processes, algorithms and systems "
    "to extract or extrapolate knowledge and insights from noisy, structured, "
    "and unstructured data. Data science is multifaceted!"
)

result = count_words(sample_paragraph)
print("Word Counts:")
for key, value in sorted(result.items()):
    print(f"  '{key}': {value}")
```
