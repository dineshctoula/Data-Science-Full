### 📖 Study Notes

#### 1. Vectorized Operations
- **Concept**: Vectorization is the process of performing mathematical operations on entire arrays at once rather than looping over individual elements.
- **Advantage**: Under the hood, NumPy delegates the loop processing to highly-optimized C/Fortran code, which runs orders of magnitude faster than standard Python `for` loops.
- **Syntax**: Operations like `+`, `-`, `*`, `/`, `**` on arrays are applied element-wise automatically.

#### 2. Broadcasting Rules
- **Concept**: Broadcasting allows NumPy to perform arithmetic operations on arrays of different shapes. The smaller array is "broadcast" across the larger array so that they have compatible shapes.
- **Broadcasting Rules**:
  1. If the arrays do not have the same rank (dimensions), prepend the shape of the lower-rank array with 1s until both shapes have the same length.
  2. Two dimensions are compatible if:
     - They are equal, or
     - One of the dimensions is exactly 1.
  3. If compatible, the arrays behave as if they had the maximum of the two shapes in each dimension.

#### 3. Universal Functions (ufuncs)
- **Concept**: Functions that operate element-wise on arrays. Examples include `np.sin()`, `np.exp()`, `np.log()`, `np.sqrt()`, and `np.add()`.

### 🧪 Practice Exercises

#### Exercise 1: Speed Comparison
Compare the time taken to multiply two arrays of size 1,000,000 using standard Python list comprehension vs NumPy vectorized multiplication.

```python
size = 1000000
arr1 = np.arange(size)
arr2 = np.arange(size)

# 1. Loop-based multiplication
start = time.time()
res_loop = [arr1[i] * arr2[i] for i in range(size)]
time_loop = time.time() - start
print(f"Loop Time: {time_loop:.5f}s")

# 2. Vectorized multiplication
start = time.time()
res_vec = arr1 * arr2
time_vec = time.time() - start
print(f"Vectorized Time: {time_vec:.5f}s")
print(f"Speedup: {time_loop / time_vec:.1f}x")
```

#### Exercise 2: Broadcasting Demo
Verify the rules of broadcasting. Create a 4x3 matrix representing daily sales for 4 different stores in 3 departments. Suppose each department has a fixed overhead cost (1x3 vector). Subtract the overhead from each store's sales using broadcasting.

```python
sales = np.array([
    [120, 80, 150],
    [200, 150, 180],
    [90, 70, 110],
    [300, 250, 400]
])
overhead = np.array([20, 15, 30])

net_profit = sales - overhead
print("Net Profit for each store per department:\n", net_profit)
```

#### Exercise 3: Practical Challenge (Matrix + Vector Broadcasting & Exp)
Create a 4x3 matrix and a 1x3 vector. Add them together to demonstrate broadcasting, then compute the exponential of the resulting matrix.

```python
# 1. Create a 4x3 matrix (values from 1 to 12)
matrix = np.arange(1, 13).reshape(4, 3)

# 2. Create a 1x3 vector
vector = np.array([10, 20, 30])

# 3. Add them using broadcasting
result_sum = matrix + vector
print("Broadcasted Sum:\n", result_sum)

# 4. Calculate exponential
result_exp = np.exp(result_sum / 10)  # scaling to avoid large values
print("\nExponential Result (Scaled):\n", np.round(result_exp, 4))
```
