### 📖 Study Notes

#### 1. What is Data Science?
Data Science is an interdisciplinary field that uses scientific methods, processes, algorithms, and systems to extract knowledge and insights from noisy, structured, and unstructured data. It lies at the intersection of:
- **Computer Science / Programming**: Writing code to clean, process, and analyze data.
- **Mathematics & Statistics**: Formulating models, validating patterns, and hypothesis testing.
- **Domain Expertise**: Applying insights to solve specific real-world/business problems.

#### 2. The Data Science Lifecycle
1. **Capture**: Data acquisition, data entry, signal reception, data extraction.
2. **Maintain**: Data warehousing, data cleansing, data staging, data processing, data architecture.
3. **Process**: Data mining, clustering/classification, data modeling, data summarization.
4. **Analyze**: Exploratory Data Analysis (EDA), predictive analysis, regression, text mining.
5. **Communicate**: Data reporting, data visualization, business intelligence, decision making.

#### 3. Basic Python Setup & Syntax
- **Variables**: Used to store data. In Python, variables are dynamically typed.
- **Print Statement**: Used to display outputs to the console.
- **f-strings**: Formatted string literals, prefixing strings with `f` or `F` and placing python expressions inside curly braces `{}`.
