### 📖 Study Notes

#### 1. Numeric Types & Basic Arithmetic
- **Integers (`int`)**: Whole numbers, positive or negative (e.g., `10`, `-3`).
- **Floats (`float`)**: Decimal numbers (e.g., `3.14`, `-0.001`).
- **Operators**:
  - Addition (`+`), Subtraction (`-`), Multiplication (`*`)
  - Division (`/`): In Python 3, standard division always returns a float (e.g., `4 / 2` is `2.0`).
  - Floor Division (`//`): Divides and truncates towards minus infinity (e.g., `17 // 5` is `3`, `-3 // 2` is `-2`).
  - Modulo (`%`): Returns the remainder of division (e.g., `17 % 5` is `2`).
  - Exponentiation (`**`): Raises base to power (e.g., `2 ** 3` is `8`).
- **Floating Point Precision**: Computers represent floats using base-2 fractions, which leads to tiny representation errors (e.g., `0.1 + 0.2` outputs `0.30000000000000004`). Use `round()` to solve this.

#### 2. Strings & Manipulation
- **Immutability**: Once created, strings cannot be modified in place. Operations return new strings.
- **Slicing**: Syntax is `string[start:stop:step]`
  - `start`: inclusive index (default `0`).
  - `stop`: exclusive index (default `len(string)`).
  - `step`: step size (default `1`, negative step reverses string).
- **Key Methods**:
  - `.strip()`: Removes leading/trailing whitespace.
  - `.upper()`, `.lower()`, `.title()`: Case transformations.
  - `.replace(old, new)`: Replace occurrences of substring.
  - `.split()`: Splits string into list of substrings.
  - `separator.join(list)`: Combines list elements with separator.
- **f-strings**: Formatted string literals (e.g., `f"{variable:.2f}"` for 2 decimal places).

#### 3. Booleans & Comparison
- **Boolean Values**: `True` and `False`.
- **Comparisons**: `==` (equal), `!=` (not equal), `<` (less than), `>` (greater than), `<=` (less or equal), `>=` (greater or equal).
- **Logical Operators**: `and` (both must be true), `or` (at least one must be true), `not` (negates value).

#### Exercise 1: Arithmetic Operations & Precision

```python
# 1. Define two variables: a = 23, b = 4
a = 23
b = 4

# 2. Calculate and print sum, product, division, floor division, modulo, and exponentiation
print(f"Sum (a + b):             {a + b}")
print(f"Product (a * b):         {a * b}")
print(f"Division (a / b):        {a / b}")
print(f"Floor Division (a // b): {a // b}")
print(f"Modulo (a % b):          {a % b}")
print(f"Exponent (a ** b):       {a ** b}")

# 3. Demonstrate floating point rounding issue and solve it
num1 = 0.1
num2 = 0.2
total = num1 + num2
print(f"\nRaw sum of 0.1 + 0.2:    {total}")
print(f"Corrected/Rounded sum:   {round(total, 2)}")
```

#### Exercise 2: String Slicing and Methods

```python
# 1. Define a string with leading/trailing whitespaces: phrase = "   learning Python is fun!  "
phrase = "   learning Python is fun!  "

# 2. Strip whitespace and make it uppercase
clean_phrase = phrase.strip().upper()
print(f"Clean & Uppercase: '{clean_phrase}'")

# 3. Extract the word "PYTHON" from your clean phrase using slicing
# Hint: "PYTHON" starts at index 9 and ends before index 15
python_word = clean_phrase[9:15]
print(f"Extracted word:    '{python_word}'")

# 4. Reverse the entire clean phrase using slicing
reversed_phrase = clean_phrase[::-1]
print(f"Reversed phrase:   '{reversed_phrase}'")
```

#### Exercise 3: Booleans and Comparisons

```python
score = 85
passing_score = 70
distinction_score = 90

# Check if score is passing and does not achieve distinction
is_passing = score >= passing_score
has_distinction = score >= distinction_score
is_average_pass = is_passing and not has_distinction

print(f"Is student passing?                      {is_passing}")
print(f"Did student achieve distinction?         {has_distinction}")
print(f"Is student passing but below distinction? {is_average_pass}")
```
