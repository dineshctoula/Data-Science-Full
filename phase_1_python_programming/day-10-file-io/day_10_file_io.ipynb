### 1. Writing and Appending to Text Files
- Mode `'w'` opens a file for writing. If the file exists, it overwrites it. If not, it creates a new one.
- Mode `'a'` opens a file for appending. It adds new text to the end of the file without deleting existing content.
- We use the `with` statement (Context Manager) because it automatically closes the file after the block finishes, releasing system resources.

```python
# 1. Writing to a new text file
with open("students.txt", "w") as file:
    file.write("Alice,85\n")
    file.write("Bob,90\n")
    print("Created students.txt and wrote initial data.")

# 2. Appending data to the same file
with open("students.txt", "a") as file:
    file.write("Charlie,95\n")
    print("Appended Charlie's data to students.txt.")
```

--- 
### 2. Reading Text Files
Python provides multiple ways to read data from a text file:
- `file.read()`: Reads the entire file as a single string.
- `file.readline()`: Reads a single line at a time.
- `file.readlines()`: Reads all lines into a list of strings.
- `for line in file`: Iterates directly over the file object (very memory efficient for large datasets).

```python
# 1. Reading the entire file at once
print("--- 1. Entire Content (read) ---")
with open("students.txt", "r") as file:
    print(file.read().strip())

# 2. Reading line by line
print("\n--- 2. Line by Line (readline) ---")
with open("students.txt", "r") as file:
    print("First line:", file.readline().strip())
    print("Second line:", file.readline().strip())

# 3. Reading lines into a list
print("\n--- 3. Lines List (readlines) ---")
with open("students.txt", "r") as file:
    lines = file.readlines()
    print([line.strip() for line in lines])

# 4. Efficient Iteration
print("\n--- 4. Memory-Efficient Iteration ---")
with open("students.txt", "r") as file:
    for line in file:
        print("Processing student:", line.strip())
```

--- 
### 3. File Exception Handling
When reading files, it's possible that the target file does not exist, leading to a `FileNotFoundError`. We can prevent our program from crashing by wrapping the file operation in a `try-except` block.

```python
# Handle FileNotFoundError gracefully
target_file = "missing_dataset.csv"
try:
    with open(target_file, "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: The file '{target_file}' could not be found.")
```

### 4. Working with JSON Files
JSON (JavaScript Object Notation) is a standard format for saving configurations, model parameters, and web API responses in data science.
- Use `json.dump()` to serialize a Python dictionary and save it to a JSON file.
- Use `json.load()` to read a JSON file and deserialize it back into a Python dictionary.

```python
import json

# Dictionary representing configuration settings
config = {
    "model_name": "Random Forest",
    "parameters": {
        "n_estimators": 100,
        "max_depth": 10
    },
    "accuracy": 0.89
}

# 1. Save dictionary to JSON file
with open("config.json", "w") as json_file:
    json.dump(config, json_file, indent=4)
    print("Saved configuration to config.json.")

# 2. Read back from JSON file
with open("config.json", "r") as json_file:
    loaded_config = json.load(json_file)
    print("Loaded Model:", loaded_config["model_name"])
    print("Loaded Accuracy:", loaded_config["accuracy"])
```

### 5. Practical Data Science Exercise
Let's write code to:
1. Create a dummy CSV dataset representing daily sales transactions. One of the records contains a corrupted/invalid non-numeric value.
2. Read the CSV file, skip the header, clean/parse the numeric fields, gracefully skip/log any corrupt rows, and calculate total sales revenue.
3. Write the final calculated sales report to a text file `sales_report.txt` and verify its output.

```python
import csv
import os

# 1. Create a raw CSV sales transaction file
raw_sales_data = [
    ["transaction_id", "product", "price", "quantity"],
    ["T1001", "Laptop", "1200", "1"],
    ["T1002", "Mouse", "25", "2"],
    ["T1003", "Keyboard", "invalid_price", "1"], # corrupt data row
    ["T1004", "Monitor", "300", "2"],
    ["T1005", "Cable", "15", "5"]
]

with open("raw_sales.csv", "w", newline="") as csv_file:
    writer = csv.writer(csv_file)
    writer.writerows(raw_sales_data)
print("Created raw_sales.csv for parsing.\n")

# 2. Read and process the CSV dataset
total_revenue = 0.0
valid_records = 0
corrupt_records = 0

with open("raw_sales.csv", "r") as csv_file:
    reader = csv.reader(csv_file)
    header = next(reader) # skip the header row
    
    for row in reader:
        tx_id, product, price_str, qty_str = row
        try:
            price = float(price_str)
            qty = int(qty_str)
            revenue = price * qty
            total_revenue += revenue
            valid_records += 1
        except ValueError:
            print(f"  [Warning] Skipping corrupt record: {row}")
            corrupt_records += 1

# 3. Write summary report to a text file
report_content = (
    "=== SALES DATA PROCESSING REPORT ===\n"
    f"Total Transactions Processed: {valid_records + corrupt_records}\n"
    f"Successful Transactions: {valid_records}\n"
    f"Corrupted/Skipped Records: {corrupt_records}\n"
    f"Total Revenue Generated: ${total_revenue:.2f}\n"
)

with open("sales_report.txt", "w") as report_file:
    report_file.write(report_content)
print("\nSummary report saved to sales_report.txt.")

# Print and verify the report content
print("\n--- Report Content ---")
print(report_content.strip())

# Clean up temporary files created during this demo
for f in ["students.txt", "config.json", "raw_sales.csv", "sales_report.txt"]:
    if os.path.exists(f):
        os.remove(f)
        print(f"Cleaned up: {f}")
```
