### 1. Function Definition & Arguments
Functions are defined using the `def` keyword. Best practices include adding type hints to parameters and return types, and writing docstrings under triple quotes `"""` to document the function's purpose and usage.

```python
# Define a function with type hints and docstring
def calculate_compound_interest(principal: float, rate: float, time: int, compounds_per_year: int = 1) -> float:
    """
    Calculates the compound interest for a given principal, rate, time, and compounding frequency.
    
    Args:
        principal (float): The initial sum of money.
        rate (float): The annual interest rate as a decimal (e.g. 0.05 for 5%).
        time (int): The time the money is invested for in years.
        compounds_per_year (int): The number of times interest is compounded per year. Default is 1.
        
    Returns:
        float: The final amount after interest.
    """
    amount = principal * (1 + rate / compounds_per_year) ** (compounds_per_year * time)
    return round(amount, 2)

# Positional arguments
amount_pos = calculate_compound_interest(1000, 0.05, 5)
print(f"Positional (1000, 5%, 5 years): ${amount_pos}")

# Keyword arguments
amount_kw = calculate_compound_interest(principal=1000, rate=0.05, time=5, compounds_per_year=12)
print(f"Keyword (1000, 5%, 5 years, compounded monthly): ${amount_kw}")

# Docstring access
print(f"\nDocstring:\n{calculate_compound_interest.__doc__}")
```

### 2. Variable-length Arguments (`*args` and `**kwargs`)
- `*args` collects any number of positional arguments into a **tuple**.
- `**kwargs` collects any number of keyword arguments into a **dictionary**.

```python
# *args Example
def calculate_mean(*args: float) -> float:
    print(f"args received: {args}, type: {type(args)}")
    if not args:
        return 0.0
    return sum(args) / len(args)

mean_val = calculate_mean(10.5, 20.0, 30.5, 40.0)
print(f"Mean: {mean_val}\n")

# **kwargs Example
def build_user_profile(first_name: str, last_name: str, **kwargs) -> dict:
    print(f"kwargs received: {kwargs}, type: {type(kwargs)}")
    profile = {
        "first_name": first_name,
        "last_name": last_name,
        **kwargs
    }
    return profile

profile = build_user_profile("Dinesh", "Toula", job="Data Scientist", country="Nepal", challenge_day=8)
print(f"Profile: {profile}")
```

### 3. Scope & the LEGB Rule
Python resolves variable names using the **LEGB** lookup rule:
1. **Local (L)**: Defined inside the current function/method.
2. **Enclosing (E)**: Defined in outer nested helper functions.
3. **Global (G)**: Defined at the module/script level.
4. **Built-in (B)**: Built-in Python names (like `len`, `sum`, `range`).

- Use `global` to modify global scope variables from a local scope.
- Use `nonlocal` to modify enclosing scope variables from a nested local scope.

```python
# Global Scope
global_variable = "Original Global"

def demonstrate_scopes():
    # Local Scope
    local_var = "I am Local"
    print(f"Local: {local_var}")
    
    def modify_global():
        global global_variable
        global_variable = "Modified Global"
        
    modify_global()
    print(f"Global after local modification: {global_variable}\n")
    
    # Enclosing & Nonlocal Scope
    def outer():
        outer_var = "Outer State"
        
        def inner():
            nonlocal outer_var
            outer_var = "Modified Outer State by Inner"
            
        inner()
        print(f"Outer variable after inner call: {outer_var}")
        
    outer()

demonstrate_scopes()
```

### 4. Lambda Expressions (Anonymous Functions)
Lambda expressions are short, one-line functions with no name. They can accept any number of arguments but only execute a single expression and return its value.

Syntax: `lambda arg1, arg2: expression`

```python
# Simple Lambda
multiply = lambda a, b: a * b
print(f"Lambda Multiply(3, 4): {multiply(3, 4)}")

# Common Use Case: Custom Sorting Key
tuples_list = [(1, "Apple", 1.50), (2, "Banana", 0.99), (3, "Cherry", 2.50)]
# Sort by price (index 2 of tuple)
sorted_list = sorted(tuples_list, key=lambda item: item[2])
print(f"Sorted by price: {sorted_list}")

# Common Use Case: map and filter
nums = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, nums))
squares = list(map(lambda x: x ** 2, nums))
print(f"Evens: {evens}")
print(f"Squares: {squares}")
```
