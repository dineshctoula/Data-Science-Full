### 📖 Study Notes

#### 1. Matrix Operations
- **Transpose**: Flipping a matrix over its diagonal, switching rows and columns. In NumPy, this is done using `A.T` or `np.transpose(A)`.
- **Matrix Multiplication**: Unlike element-wise multiplication (`*`), matrix multiplication computes dot products of rows and columns. In NumPy, use the `@` operator or `np.dot(A, B)`.

#### 2. Determinants & Matrix Inverses
- **Identity Matrix**: A square matrix with ones on the main diagonal and zeros elsewhere, created using `np.eye(n)`.
- **Determinant**: A scalar value that describes the scaling factor of the linear transformation described by the matrix. Computed using `np.linalg.det(A)`.
- **Matrix Inverse**: A matrix $A^{-1}$ such that $A \cdot A^{-1} = I$ (Identity Matrix). Computed using `np.linalg.inv(A)`. Only non-singular (determinant $\neq 0$) square matrices have an inverse.

#### 3. Solving Linear Systems
- A system of linear equations can be represented in matrix form as:
  $$Ax = B$$
  where:
  - $A$ is the coefficient matrix.
  - $x$ is the vector of unknown variables.
  - $B$ is the constants vector.
- Rather than computing $x = A^{-1}B$ (which is computationally expensive and numerically unstable), NumPy provides a optimized solver: `np.linalg.solve(A, B)`.

### 🧪 Practice Exercises

#### Exercise 1: Matrix Multiplication & Transpose
Define a 2x3 matrix $A$ and a 3x2 matrix $B$. Compute the transpose of $A$, and multiply $A$ by $B$ to get a 2x2 matrix.

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
B = np.array([
    [7, 8],
    [9, 10],
    [11, 12]
])

print("A^T:\n", A.T)
print("\nA @ B:\n", A @ B)
```

#### Exercise 2: Determinant and Inverse
Given a square matrix $M$, calculate its determinant. If the determinant is non-zero, calculate its inverse and verify that $M \cdot M^{-1} = I$.

```python
M = np.array([
    [3, 4],
    [2, 3]
])

det_M = np.linalg.det(M)
print(f"Determinant of M: {det_M:.2f}")

if not np.isclose(det_M, 0):
    M_inv = np.linalg.inv(M)
    print("\nInverse of M:\n", M_inv)
    print("\nVerification (M @ M^-1):\n", np.round(M @ M_inv, 2))
else:
    print("Matrix is singular (not invertible).")
```

#### Exercise 3: Practical Challenge (Solving Systems of Linear Equations)
Solve the system of equations:
$$2x + y = 5$$
$$x - 3y = -1$$
using `np.linalg.solve`.

```python
# Coefficients Matrix A
A = np.array([
    [2, 1],
    [1, -3]
])

# Constants Vector B
B = np.array([5, -1])

# Solve Ax = B
solution = np.linalg.solve(A, B)

print(f"Solution: x = {solution[0]:.4f}, y = {solution[1]:.4f}")

# Verify result
print("Verification check:")
print("  2x + y =", 2 * solution[0] + solution[1])
print("  x - 3y =", solution[0] - 3 * solution[1])
```
